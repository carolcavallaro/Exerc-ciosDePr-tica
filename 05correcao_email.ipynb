--- EXERCÍCIO DE PRÁTICA: DESAFIO PROCESSADOR DE E-MAILS E USUÁRIOS ---

Crie um script Python chamado processador_email.py que faça o seguinte:
1. Receba do usuário uma string contendo um e-mail completo (exemplo: usuario@gmail.com).
2. Extraia o nome do usuário (tudo o que vem antes do @).
3. Extraia o domínio (tudo o que vem depois do @).
4. Verifique e exiba se o e-mail termina com .com ou .br utilizando índice negativo ([-4:] ou [-3:]).
5. Imprima o nome do usuário invertido e em letras maiúsculas.

```python
# --- MEU CÓDIGO ---

email = input("Digite o seu e-mail completo: ")
nomeUsuario = email.split('@')[0] # .split('@') para dividir a frase a partir do @ | e [0] para dar nome ao usuário do começo ao fim dessa divisão. Como o usuário vem antes do @, não precisa mais nada antes.
dominioEmail = email.split('@')[1] # [-1] também funciona
fimEmail = dominioEmail.split('.')[-1]
invertido = nomeUsuario.upper()[::-1] # .upper() para letras maiúsculas
print(f"Seu usuário é: {nomeUsuario} \nSeu domínio é: @{dominioEmail} \nSeu e-mail termina com: .{fimEmail} \nSeu nome de usuário invertido fica: {invertido}")

# --- FIM DO ALGORITMO ---
```

```python
# --- COREÇÃO GEMINI ---

# 1. Recebe o e-mail completo do usuário
email = input("Digite o seu e-mail completo: ").strip()

# 2. Extrai o nome do usuário (tudo antes do @)
nomeUsuario = email.split('@')[0]

# 3. Extrai o domínio (tudo depois do @)
dominioEmail = email.split('@')[1]

# 4. Verifica se termina com .com ou .br usando índices negativos
# [-4:] pega os últimos 4 caracteres (ex: '.com')
# [-3:] pega os últimos 3 caracteres (ex: '.br')
eh_com = email[-4:] == ".com"
eh_br = email[-3:] == ".br"

# 5. Inverte o nome do usuário e coloca em maiúsculas
invertido = nomeUsuario.upper()[::-1]

# Exibição dos resultados
print(f"\n--- RESULTADOS ---")
print(f"Nome de usuário: {nomeUsuario}")
print(f"Domínio: {dominioEmail}")

# Validação do item 4
if eh_com:
    print("O e-mail termina com: .com")
elif eh_br:
    print("O e-mail termina com: .br")
else:
    print("O e-mail não termina com .com nem com .br")

print(f"Usuário invertido (maiúsculo): {invertido}")
```

O que mudou e por quê?
- email.split('@')[1] no domínio: Usar o índice [1] em vez de [-1] é mais seguro. Se o usuário digitasse algo inválido com mais de um @, o [1] garante que pegamos exatamente a parte logo após o primeiro @.
- Uso de email[-4:] e email[-3:]: O fatiamento (slicing) no Python funciona assim:
   - email[-4:] lê a string do quarto caractere de trás para frente até o final. Para "usuario@gmail.com", isso resulta em ".com".
   - email[-3:] lê do terceiro caractere de trás para frente até o final. Para "usuario@site.com.br", isso resulta em ".br".
- Uso de .strip() no input(): É uma boa prática para remover acidentalmente espaços em branco que o usuário possa digitar antes ou depois do e-mail.
